# chatbot_rag API 테스트

FastAPI 서버를 실행한 뒤 Python `requests`로 기본 API를 확인한다.

서버 실행:

```bash
python -m uvicorn app.main:app --reload
```

````python
import requests

BASE_URL = "http://localhost:8000"
````

## Health check

````python
response = requests.get(f"{BASE_URL}/health", timeout=10)
response.raise_for_status()
response.json()
````

## 문서 색인 상태

````python
response = requests.get(f"{BASE_URL}/documents/stats", timeout=10)
response.raise_for_status()
response.json()
````

## RAG 질문

````python
payload = {"message": "환불은 언제까지 가능한가요?"}
response = requests.post(f"{BASE_URL}/chat", json=payload, timeout=180)
response.raise_for_status()
response.json()
````
